```python
import numpy as np                  # tool for working with numbers and lists
import cv2                          # tool for reading the image
import matplotlib.pyplot as plt     # tool for drawing graphs

img = cv2.imread("/content/input.jpg", cv2.IMREAD_GRAYSCALE)   # open the picture in black and white
L = 256                             # 256 brightness values (0 to 255)
M, N = img.shape                    # M = height of the picture, N = width of the picture

hist = np.zeros(L)                  # 256 empty boxes, all starting at 0
for v in img.ravel():               # take every pixel one by one
    hist[v] += 1                    # add 1 to the box of that brightness

pdf = hist / (M * N)                # divide each count by total pixels, so counts become fractions
                                    # example: 500 pixels out of 10000 becomes 0.05

print("Total =", pdf.sum())         # all fractions together should make 1

plt.bar(np.arange(L), pdf, width=1.0, color='steelblue')   # draw one bar for each fraction
plt.title('Normalised Histogram (PDF)')    # title of the graph
plt.xlabel('Brightness')            # bottom line shows brightness
plt.ylabel('Fraction of pixels')    # side line shows the fraction (between 0 and 1)
plt.show()                          # show the graph
```
