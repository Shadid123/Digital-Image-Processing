```python
import numpy as np                  # tool for working with numbers and lists
import cv2                          # tool for reading the image
import matplotlib.pyplot as plt     # tool for drawing graphs

img = cv2.imread("/content/input.jpg", cv2.IMREAD_GRAYSCALE)   # open the picture in black and white
L = 256                             # brightness can be 0 to 255, so 256 possible values

hist = np.zeros(L)                  # make 256 empty boxes, all starting at 0

for v in img.ravel():               # take every pixel of the picture one by one
    hist[v] += 1                    # put it in the box of its brightness (add 1 to that box)

plt.bar(np.arange(L), hist, width=1.0, color='gray')   # draw one bar for each box
plt.title('Histogram')              # title of the graph
plt.xlabel('Brightness')            # bottom line shows brightness 0 to 255
plt.ylabel('Number of pixels')      # side line shows how many pixels have that brightness
plt.show()                          # show the graph
```
