```python

import cv2  # Used for image processing and affine transformations
import numpy as np  # Used to create transformation matrices
import os  # Used to create the output folder and manage file paths
import matplotlib.pyplot as plt  # Used to display images and results

# Read the input image
image = cv2.imread("/content/input.jpg")

# Check whether the image was loaded successfully
if image is None:
    raise ValueError("Image not found. Check the filename.")

# Convert BGR to RGB for correct display in Matplotlib
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Get the height and width of the image
height, width = image.shape[:2]

# Create the output folder if it does not exist
os.makedirs("/content/output", exist_ok=True)

# Apply translation: move the image 50 pixels right and 30 pixels down
translation_matrix = np.float32([
    [1, 0, 50],
    [0, 1, 30]
])

translated = cv2.warpAffine(image, translation_matrix, (width, height))

# Apply rotation: rotate the image 45 degrees around its center
rotation_matrix = cv2.getRotationMatrix2D(
    (width // 2, height // 2), 45, 1
)

rotated = cv2.warpAffine(image, rotation_matrix, (width, height))

# Apply scaling: reduce the image to 80% of its original size
scaling_matrix = np.float32([
    [0.8, 0, 0],
    [0, 0.8, 0]
])

scaled = cv2.warpAffine(image, scaling_matrix, (width, height))

# Save the transformed images after converting RGB back to BGR
cv2.imwrite("/content/output/translated.jpg",
            cv2.cvtColor(translated, cv2.COLOR_RGB2BGR))

cv2.imwrite("/content/output/rotated.jpg",
            cv2.cvtColor(rotated, cv2.COLOR_RGB2BGR))

cv2.imwrite("/content/output/scaled.jpg",
            cv2.cvtColor(scaled, cv2.COLOR_RGB2BGR))

# Display the original and transformed images
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(translated)
plt.title("Translation")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(rotated)
plt.title("Rotation")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(scaled)
plt.title("Scaling")
plt.axis("off")

# Adjust spacing and display the results
plt.tight_layout()
plt.show()

print("Affine transformation completed.")
print("Images saved in the output folder.")
```
