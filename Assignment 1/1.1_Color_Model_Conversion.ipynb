```python

import cv2  # Used for image processing and color model conversion
import os  # Used to create the output folder and manage file paths
import matplotlib.pyplot as plt  # Used to display images and results

# Read the input image
image = cv2.imread("/content/input.jpg")

# Check whether the image was loaded successfully
if image is None:
    raise ValueError("Image not found. Check the filename.")

# Convert BGR to RGB for correct display in Matplotlib
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert the image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Convert the image from BGR to HSV
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# Convert HSV back to RGB for displaying its colors correctly
hsv_display = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

# Create the output folder if it does not exist
os.makedirs("/content/output", exist_ok=True)

# Save the converted images
cv2.imwrite("/content/output/rgb.jpg", cv2.cvtColor(rgb, cv2.COLOR_RGB2BGR))
cv2.imwrite("/content/output/grayscale.jpg", gray)
cv2.imwrite("/content/output/hsv.jpg", hsv)

# Display the original and converted images
plt.figure(figsize=(12, 4))

plt.subplot(1, 4, 1)
plt.imshow(rgb)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(hsv_display)
plt.title("HSV Display")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(hsv)
plt.title("HSV Channels")
plt.axis("off")

# Adjust spacing and display the results
plt.tight_layout()
plt.show()

print("Color model conversion completed.")
print("Images saved in the output folder.")
```
