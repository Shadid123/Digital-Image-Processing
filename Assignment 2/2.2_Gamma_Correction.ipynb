```python

import cv2  # Used for image processing
import numpy as np  # Used for mathematical operations on pixel values
import os  # Used to create folders and manage file paths
import matplotlib.pyplot as plt  # Used to display the output image

# Read the input image from Google Colab
image = cv2.imread("/content/input.jpg")

# Check whether the image was loaded successfully
if image is None:
    raise ValueError("Image not found. Check the filename.")

# Convert the color image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Set the gamma value; values below 1 brighten the image
gamma = 0.5

# Normalize pixel values from the range 0–255 to 0–1
normalized = gray / 255.0

# Apply the gamma correction formula to each pixel
corrected = np.power(normalized, gamma)

# Convert pixel values back to the range 0–255
corrected = np.uint8(corrected * 255)

# Create the output folder if it does not already exist
os.makedirs("/content/output", exist_ok=True)

# Save the gamma-corrected image
cv2.imwrite("/content/output/gamma_correction.jpg", corrected)

# Display the corrected image using Matplotlib
plt.imshow(corrected, cmap="gray")
plt.title("Gamma Correction")
plt.axis("off")  # Hide the axis numbers
plt.show()

# Print completion messages
print("Gamma correction completed.")
print("Image saved in the output folder.")
```
