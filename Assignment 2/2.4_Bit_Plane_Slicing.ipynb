```python

import cv2  # Used for image processing and reading/writing images
import numpy as np  # Used for numerical operations on image pixel values
import os  # Used to create folders and manage file paths
import matplotlib.pyplot as plt  # Used to display images and bit planes

# Read the input image
image = cv2.imread("/content/input.jpg")

# Check whether the image was loaded successfully
if image is None:
    raise ValueError("Image not found. Check the filename.")

# Convert the color image into grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Create the output folder if it does not exist
os.makedirs("/content/output", exist_ok=True)

# Create a figure to display all 8 bit planes
plt.figure(figsize=(12, 6))

# Extract bit planes from bit 0 to bit 7
for i in range(8):

    # Shift the selected bit to the rightmost position
    # Then use & 1 to extract that bit from every pixel
    bit_plane = (gray >> i) & 1

    # Convert bit values 0 and 1 into black and white (0 and 255)
    bit_plane = bit_plane * 255

    # Save the extracted bit plane as an image
    cv2.imwrite(
        f"/content/output/bit_plane_{i}.jpg",
        bit_plane
    )

    # Display the current bit plane
    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_plane, cmap="gray")
    plt.title(f"Bit Plane {i}")
    plt.axis("off")

# Adjust spacing between the displayed images
plt.tight_layout()

# Show all 8 bit planes
plt.show()

print("Bit plane slicing completed.")
print("All bit planes saved in the output folder.")
```
