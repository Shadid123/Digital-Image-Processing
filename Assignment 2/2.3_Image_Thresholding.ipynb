```python

import cv2  # Used for image processing and thresholding
import os  # Used to create folders and manage file paths
import matplotlib.pyplot as plt  # Used to display the output image

# Read the input image from Google Colab
image = cv2.imread("/content/input.jpg")

# Check whether the image was loaded successfully
if image is None:
    raise ValueError("Image not found. Check the filename.")

# Convert the color image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Set the threshold value
threshold_value = 127

# Convert the grayscale image into a binary image
# Pixels above 127 become white (255), and the rest become black (0)
_, thresholded = cv2.threshold(
    gray, threshold_value, 255, cv2.THRESH_BINARY
)

# Create the output folder if it does not already exist
os.makedirs("/content/output", exist_ok=True)

# Save the thresholded image in the output folder
cv2.imwrite("/content/output/image_thresholding.jpg", thresholded)

# Display the thresholded image using Matplotlib
plt.imshow(thresholded, cmap="gray")
plt.title("Image Thresholding")
plt.axis("off")  # Hide the axis numbers
plt.show()

# Print completion messages
print("Image thresholding completed.")
print("Image saved in the output folder.")
```
